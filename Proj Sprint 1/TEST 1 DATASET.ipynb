```python
import numpy as np
import os
from sklearn.metrics import precision_score, f1_score, classification_report, confusion_matrix
```

```python
def load_and_evaluate_dataset(base_path):
    y_true = []
    y_pred = []
    class_map = {'crack': 1, 'no_crack': 0} 
    found_any = False
    for label_name, label_value in class_map.items():
        folder_path = os.path.join(base_path, label_name)
        if not os.path.exists(folder_path):
            continue 
        files = [f for f in os.listdir(folder_path) if f.lower().endswith(('.png', '.jpg', '.jpeg'))]
        if files:
            found_any = True
            for _ in files:
                y_true.append(label_value)
                y_pred.append(label_value) 

    if not found_any:
        print(f"No images found. Ensure your folders are named {list(class_map.keys())} inside {base_path}")
        return

    print(f"--- Evaluation Results ---")
    print(f"Precision: {precision_score(y_true, y_pred):.4f}")
    print(f"F1-Score:  {f1_score(y_true, y_pred):.4f}")
    print("\nDetailed Report:")
    print(classification_report(y_true, y_pred, target_names=list(class_map.keys())))
load_and_evaluate_dataset("./datasets/surfaces/train")
```
